## 1. Introduction to Pandas

### What is Pandas?

**Pandas** is Python's premier data manipulation library, built on top of NumPy.

| Aspect | Description |
|--------|-------------|
| **Name Origin** | "Panel Data" - econometrics term for multidimensional structured data |
| **Creator** | Wes McKinney (2008) |
| **Primary Use** | Data wrangling, cleaning, analysis |
| **Key Strength** | Intuitive API for tabular data with mixed types |

### Why Pandas for ML?

- **Real-world data is messy**: Missing values, mixed types, inconsistent formats
- **Feature engineering**: Creating ML features requires data manipulation
- **EDA (Exploratory Data Analysis)**: Understanding data before modeling
- **Integration**: Works seamlessly with scikit-learn, NumPy, matplotlib

## 2. Series and DataFrames

### 2.1 Series: 1-Dimensional Labeled Array

A **Series** is like a column in a spreadsheet - a 1D array with an index.

```
Index    Values
  0   →   10
  1   →   20
  2   →   30
```

````python
# Creating a Series from a list
values = [10, 20, 30, 40, 50]
s = pd.Series(values)
print("Basic Series:")
print(s)
print(f"\nType: {type(s)}")
print(f"Index: {s.index.tolist()}")
print(f"Values: {s.values}")
````

````python
# Series with custom index (labeled)
temperatures = pd.Series(
    [72.5, 75.2, 68.9, 71.3, 74.8],
    index=["Mon", "Tue", "Wed", "Thu", "Fri"],
    name="Temperature (°F)",
)
print("Labeled Series:")
print(temperatures)
print(f"\nAccess by label: temperatures['Wed'] = {temperatures['Wed']}")
print(f"Access by position: temperatures.iloc[2] = {temperatures.iloc[2]}")
````

````python
# Series operations are vectorized (like NumPy)
print("Vectorized Operations:")
print(f"Mean temperature: {temperatures.mean():.1f}°F")
print(f"Max temperature: {temperatures.max():.1f}°F on {temperatures.idxmax()}")
print(f"\nTemperatures > 72°F:")
print(temperatures[temperatures > 72])
````

### 2.2 DataFrame: 2-Dimensional Labeled Table

A **DataFrame** is like a spreadsheet or SQL table - rows and columns with labels.

```
         col_A    col_B    col_C
row_0      1       'a'      True
row_1      2       'b'      False
row_2      3       'c'      True
```

**Key characteristics:**
- Each column is a **Series**
- Columns can have different data types
- Has both row index and column index

````python
# ============================================================
# CREATING DATAFRAMES - The Starting Point of Every ML Project
# ============================================================
# DataFrames are your primary data structure for:
# - Loading datasets (CSV, Excel, SQL)
# - Feature engineering and transformation
# - Exploratory Data Analysis (EDA)
# - Preparing data for sklearn models
#
# Key concept: Each ROW is a sample, each COLUMN is a feature

data = {
    "Name": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
    "Age": [25, 30, 35, 28, 32],
    "City": ["NYC", "LA", "Chicago", "NYC", "LA"],
    "Salary": [70000, 80000, 90000, 75000, 85000],
    "Experience": [2, 5, 8, 4, 6],
}

df = pd.DataFrame(data)
print("DataFrame from dictionary:")
print(df)
````

````python
# DataFrame info and inspection
print("=" * 50)
print("DataFrame Info:")
print("=" * 50)
print(f"Shape: {df.shape} (rows, columns)")
print(f"Columns: {df.columns.tolist()}")
print(f"Index: {df.index.tolist()}")
print(f"\nData types:\n{df.dtypes}")
````

````python
# Quick data exploration
print("First 2 rows:")
print(df.head(2))
print("\nLast 2 rows:")
print(df.tail(2))
print("\nStatistical summary:")
print(df.describe())
````

````python
# Accessing columns
print("Access single column (returns Series):")
print(df["Name"])
print(f"\nType: {type(df['Name'])}")

print("\nAccess multiple columns (returns DataFrame):")
print(df[["Name", "Salary"]])
````

### 2.3 Creating DataFrames - Multiple Methods

````python
# Method 1: From list of dictionaries (row-oriented)
records = [
    {"product": "Laptop", "price": 999, "stock": 50},
    {"product": "Phone", "price": 699, "stock": 150},
    {"product": "Tablet", "price": 499, "stock": 75},
]
df_records = pd.DataFrame(records)
print("From list of dicts:")
print(df_records)
````

````python
# Method 2: From NumPy array
np_array = np.random.randn(4, 3)  # 4 rows, 3 columns of random numbers
df_numpy = pd.DataFrame(
    np_array,
    columns=["Feature_A", "Feature_B", "Feature_C"],
    index=["Sample_1", "Sample_2", "Sample_3", "Sample_4"],
)
print("From NumPy array:")
print(df_numpy)
````

````python
# Method 3: From Series
s1 = pd.Series([1, 2, 3], name="A")
s2 = pd.Series([4, 5, 6], name="B")
df_series = pd.concat([s1, s2], axis=1)
print("From Series:")
print(df_series)
````

## 4. Data Selection and Indexing

### The Golden Rule: `loc` vs `iloc`

| Method | Selection By | Example |
|--------|--------------|--------|
| `loc` | **Labels** (names) | `df.loc['row_a', 'col_x']` |
| `iloc` | **Integer positions** | `df.iloc[0, 2]` |
| `[]` | Columns or boolean | `df['col']` or `df[df['col'] > 5]` |

**Remember**: `loc` = **L**abel, `iloc` = **I**nteger position

| `df.iloc[:, -1]` | Select target column |

### Why This Matters for ML:| `df.iloc[:, :-1]` | Select all features (exclude target) |

| `df[df['label'] == 1]` | Filter positive class samples |

| Operation | ML Use Case || `df.loc[:, features]` | Select specific feature columns |

|-----------|-------------|| `df.iloc[train_idx]` | Select training samples by index |

````python
# Create a DataFrame with custom index for demonstration
df_indexed = pd.DataFrame(
    {
        "Name": ["Alice", "Bob", "Charlie", "Diana", "Eve"],
        "Age": [25, 30, 35, 28, 32],
        "City": ["NYC", "LA", "Chicago", "NYC", "LA"],
        "Salary": [70000, 80000, 90000, 75000, 85000],
    },
    index=["emp_001", "emp_002", "emp_003", "emp_004", "emp_005"],
)

print("DataFrame with custom index:")
print(df_indexed)
````

````python
# loc: Selection by LABEL
print("=== loc: Selection by Label ===\n")

# Single value
print(f"df.loc['emp_002', 'Name'] = {df_indexed.loc['emp_002', 'Name']}")

# Single row (returns Series)
print(f"\ndf.loc['emp_003']:\n{df_indexed.loc['emp_003']}")

# Multiple rows
print(f"\ndf.loc[['emp_001', 'emp_003']]:\n{df_indexed.loc[['emp_001', 'emp_003']]}")

# Rows and columns
print(f"\ndf.loc['emp_001':'emp_003', 'Name':'City']:")
print(df_indexed.loc["emp_001":"emp_003", "Name":"City"])
````

````python
# iloc: Selection by INTEGER POSITION
print("=== iloc: Selection by Position ===\n")

# Single value (row 1, column 0)
print(f"df.iloc[1, 0] = {df_indexed.iloc[1, 0]}")

# Single row
print(f"\ndf.iloc[2]:\n{df_indexed.iloc[2]}")

# Slice rows (0 to 2, exclusive of 3)
print(f"\ndf.iloc[0:3]:")
print(df_indexed.iloc[0:3])

# Rows and columns by position
print(f"\ndf.iloc[0:2, 0:2]:")
print(df_indexed.iloc[0:2, 0:2])
````

````python
# ============================================================
# BOOLEAN INDEXING - CRUCIAL for ML Data Preparation!
# ============================================================
# This is one of the most important tools for:
# 1. Filtering outliers before training
# 2. Selecting specific class samples
# 3. Creating train/validation splits by condition
# 4. Removing missing or invalid data
# 5. Stratified sampling by category

print("=== Boolean Indexing ===\n")

# Filter rows where Age > 28
mask = df_indexed["Age"] > 28
print(f"Mask (Age > 28):\n{mask}\n")
print(f"Filtered rows:\n{df_indexed[mask]}")
````

````python
# ============================================================
# COMBINING CONDITIONS - Essential for Feature Engineering
# ============================================================
# In ML, you often need to combine multiple conditions:
# - Select samples meeting multiple criteria
# - Create binary features from conditions
# - Filter training data by multiple rules
#
# IMPORTANT: Use parentheses around each condition!
# & = AND, | = OR, ~ = NOT

# Age > 25 AND Salary > 75000
high_earners = df_indexed[(df_indexed["Age"] > 25) & (df_indexed["Salary"] > 75000)]
print("Age > 25 AND Salary > 75000:")
print(high_earners)

# City is NYC OR LA
coastal = df_indexed[(df_indexed["City"] == "NYC") | (df_indexed["City"] == "LA")]
print("\nCity is NYC or LA:")
print(coastal)
````

````python
# Using isin() for multiple values
cities_of_interest = ["NYC", "Chicago"]
filtered = df_indexed[df_indexed["City"].isin(cities_of_interest)]
print(f"Employees in {cities_of_interest}:")
print(filtered)

# Using query() for SQL-like filtering
result = df_indexed.query("Age > 28 and Salary > 75000")
print("\nUsing query():")
print(result)
````
