```python
# Solution
from collections import defaultdict

def group_by_solution(items: List[dict], key: str) -> Dict[str, List[dict]]:
    grouped = defaultdict(list)
    for item in items:
        grouped[item[key]].append(item)
    return dict(grouped)

students = [
    {'name': 'Alice', 'grade': 'A'},
    {'name': 'Bob', 'grade': 'B'},
    {'name': 'Charlie', 'grade': 'A'},
    {'name': 'Diana', 'grade': 'B'},
]
for grade, students_list in group_by_solution(students, 'grade').items():
    print(f"Grade {grade}: {[s['name'] for s in students_list]}")
```

Create a decorator that caches function results.

**Difficulty:** Medium | **Skill:** Decorators, Closures

```python
import time

def cache(func):
    """
    A decorator that caches function results.
    
    Returns cached result if function called with same arguments.
    """
    # YOUR CODE HERE
    # Hint: Use a dictionary to store results, key = args
    pass


# Test with Fibonacci (will be slow without caching)
@cache
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n - 1) + fibonacci(n - 2)

# This should be fast due to caching
start = time.time()
result = fibonacci(30)
elapsed = time.time() - start
print(f"fibonacci(30) = {result}, took {elapsed:.4f} seconds")
assert elapsed < 1.0, "Cache not working - should be instant!"
print("✅ Test passed!")
```

```python
# Solution
from functools import wraps

def cache_solution(func):
    memo = {}
    @wraps(func)
    def wrapper(*args):
        if args not in memo:
            memo[args] = func(*args)
        return memo[args]
    return wrapper

@cache_solution
def fib_cached(n):
    if n < 2:
        return n
    return fib_cached(n - 1) + fib_cached(n - 2)

start = time.time()
print(f"fib(35) = {fib_cached(35)}, took {time.time() - start:.4f}s")
```

```python
# Solution
from functools import reduce as functools_reduce

class DataPipelineSolution:
    def __init__(self, data):
        self.data = list(data)
    
    def filter(self, func):
        self.data = [x for x in self.data if func(x)]
        return self
    
    def map(self, func):
        self.data = [func(x) for x in self.data]
        return self
    
    def reduce(self, func, initial=None):
        if initial is not None:
            return functools_reduce(func, self.data, initial)
        return functools_reduce(func, self.data)
    
    def collect(self):
        return self.data

# Demo
result = DataPipelineSolution(range(10)).filter(lambda x: x % 2 == 0).map(lambda x: x ** 2).collect()
print(f"Even squares from 0-9: {result}")

total = DataPipelineSolution([1, 2, 3, 4, 5]).reduce(lambda a, b: a + b)
print(f"Sum of 1-5: {total}")
```

```python
# Solution
def is_valid_parentheses_solution(s: str) -> bool:
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    
    for char in s:
        if char in mapping:  # closing bracket
            if not stack or stack.pop() != mapping[char]:
                return False
        else:  # opening bracket
            stack.append(char)
    
    return len(stack) == 0

print(f"'()[]{{}}': {is_valid_parentheses_solution('()[]{}')}")  # True
print(f"'([)]': {is_valid_parentheses_solution('([)]')}")  # False
```

## Exercise 8: Merge Intervals
Given a list of intervals, merge overlapping ones.

**Example:** `[[1,3],[2,6],[8,10],[15,18]]` → `[[1,6],[8,10],[15,18]]`

**Difficulty:** Medium | **Common in:** Real-world scheduling problems

```python
def merge_intervals(intervals: List[List[int]]) -> List[List[int]]:
    """
    Merge overlapping intervals.
    
    Args:
        intervals: List of [start, end] pairs
    
    Returns:
        List of merged intervals
    """
    # YOUR CODE HERE
    # Hint: Sort by start time, then iterate
    pass


# Test cases
assert merge_intervals([[1,3],[2,6],[8,10],[15,18]]) == [[1,6],[8,10],[15,18]]
assert merge_intervals([[1,4],[4,5]]) == [[1,5]]
print("✅ All tests passed!")
```

```python
# Solution
def merge_intervals_solution(intervals: List[List[int]]) -> List[List[int]]:
    if not intervals:
        return []
    
    # Sort by start time
    intervals.sort(key=lambda x: x[0])
    
    merged = [intervals[0]]
    
    for start, end in intervals[1:]:
        # If current interval overlaps with last merged
        if start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    
    return merged

print(merge_intervals_solution([[1,3],[2,6],[8,10],[15,18]]))
```
