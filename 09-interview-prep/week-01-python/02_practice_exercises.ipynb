## 📚 Exercise 1: Two Sum - Hash Map Pattern

### What is a Hash Map?
A **hash map** (also called dictionary in Python) is a data structure that stores key-value pairs with **O(1) average-time complexity** for lookups, insertions, and deletions.

### How Does It Work Under the Hood?
1. **Hashing**: When you insert a key, Python computes a hash code using the `hash()` function
2. **Bucket Assignment**: The hash code is used to determine which "bucket" (memory location) to store the value
3. **Collision Handling**: If two keys hash to the same bucket, Python uses "open addressing" to find the next available slot

```python
# Example of hashing
hash('hello')  # Returns an integer like -1234567890
```

### Why Use Hash Maps for Two Sum?
- **Brute Force**: Check every pair → O(n²) time complexity
- **Hash Map**: Store seen values, check for complement → O(n) time complexity

The key insight: For each number `num`, we need to find if `target - num` exists. A hash map answers "have I seen this value before?" in O(1) time.

### Real-World Applications
- **Database Indexing**: Finding records by key
- **Caching**: Storing computed results (memoization)
- **Counting/Grouping**: Word frequencies, user sessions
- **Deduplication**: Checking if item already processed

````python
# Solution with detailed walkthrough
def two_sum_solution(nums: list, target: int) -> list:
    seen = {}  # Hash map: value -> index
    
    for i, num in enumerate(nums):
        complement = target - num
        
        # Check if we've seen the complement before
        if complement in seen:  # O(1) lookup!
            return [seen[complement], i]
        
        # Store current number for future lookups
        seen[num] = i
    
    return []  # No solution found

# Trace through example: nums=[2,7,11,15], target=9
print("Tracing two_sum([2, 7, 11, 15], 9):")
print("  i=0, num=2: complement=7, seen={} → not found, add {2: 0}")
print("  i=1, num=7: complement=2, seen={2: 0} → FOUND! Return [0, 1]")
print(f"\nResult: {two_sum_solution([2, 7, 11, 15], 9)}")
````

## 📚 Exercise 3: Flatten Nested List - Recursion Deep Dive

### What is Recursion?
**Recursion** is when a function calls itself to solve smaller instances of the same problem. Every recursive solution has:
1. **Base Case**: When to stop (no more nested lists)
2. **Recursive Case**: How to break down the problem

### How Does Recursion Work Under the Hood?
Python uses a **call stack** to manage function calls:

```
flatten([[1, [2, 3]], 4])
├── item=[[1, [2, 3]]] → is list, recurse
│   ├── item=1 → not list, append 1
│   └── item=[2, 3] → is list, recurse
│       ├── item=2 → append 2
│       └── item=3 → append 3
└── item=4 → not list, append 4
Result: [1, 2, 3, 4]
```

### Why Use Recursion for Nested Structures?
- **Natural Fit**: Tree-like data (JSON, XML, file systems)
- **Unknown Depth**: Don't know how deep nesting goes
- **Divide & Conquer**: Complex problems become simple subproblems

### Recursion vs Iteration Trade-offs
| Aspect | Recursion | Iteration |
|--------|-----------|----------|
| Readability | More intuitive for trees | Better for linear data |
| Memory | Uses call stack (limited) | Uses explicit stack |
| Speed | Slight overhead | Usually faster |
| Python Limit | ~1000 calls by default | No limit |

### Real-World Applications
- **JSON/XML Parsing**: Traversing nested structures
- **File System Operations**: Walking directories
- **Tree Algorithms**: Binary search trees, DOM manipulation
- **Graph Traversal**: DFS (Depth-First Search)

---

````python
# Solution with trace
def flatten_solution(nested_list: list, depth=0) -> list:
    result = []
    indent = "  " * depth
    
    for item in nested_list:
        if isinstance(item, list):
            print(f"{indent}Found list: {item} → recursing...")
            result.extend(flatten_solution(item, depth + 1))
        else:
            print(f"{indent}Found element: {item} → appending")
            result.append(item)
    
    return result

# Trace execution
print("Tracing flatten([[1, 2], [3, [4, 5]], 6]):")
result = flatten_solution([[1, 2], [3, [4, 5]], 6])
print(f"\nFinal result: {result}")

# Also show iterative approach for comparison
def flatten_iterative(nested_list: list) -> list:
    """Iterative version using explicit stack"""
    stack = list(reversed(nested_list))  # Use list as stack
    result = []
    
    while stack:
        item = stack.pop()
        if isinstance(item, list):
            stack.extend(reversed(item))  # Add to stack
        else:
            result.append(item)
    
    return result

print(f"\nIterative result: {flatten_iterative([[1, 2], [3, [4, 5]], 6])}")
````

## 📚 Exercise 5: Cache Decorator (Memoization)

### What is Memoization?
**Memoization** is an optimization technique that stores the results of expensive function calls and returns the cached result when the same inputs occur again.

### How Do Decorators Work?
A decorator is a function that wraps another function, adding behavior:

```python
@my_decorator
def my_function():
    pass

# Is equivalent to:
def my_function():
    pass
my_function = my_decorator(my_function)
```

### Decorator Anatomy
```python
def decorator(func):           # Takes a function
    cache = {}                  # Closure variable (persists between calls)
    
    def wrapper(*args):         # Replacement function
        if args in cache:
            return cache[args]  # Return cached result
        result = func(*args)    # Call original function
        cache[args] = result    # Store result
        return result
    
    return wrapper              # Return the wrapper
```

### Why Use Memoization?
- **Overlapping Subproblems**: Same computation repeated (Fibonacci, dynamic programming)
- **Expensive Computations**: API calls, database queries, complex calculations
- **Trade-off**: Memory for speed

### Python's Built-in: @functools.lru_cache
```python
from functools import lru_cache

@lru_cache(maxsize=128)  # LRU = Least Recently Used
def fibonacci(n):
    if n < 2:
        return n
    return fibonacci(n-1) + fibonacci(n-2)
```

### Real-World Applications
- **Web Caching**: Redis, Memcached for session data
- **API Rate Limiting**: Cache responses to reduce API calls
- **Database Query Caching**: ORM query results
- **Dynamic Programming**: All DP problems benefit from memoization

---

````python
# Solution with detailed explanation
def cache_solution(func):
    memo = {}  # Closure: persists between function calls
    
    @wraps(func)  # Preserves __name__, __doc__, etc.
    def wrapper(*args):
        if args not in memo:
            memo[args] = func(*args)  # Compute and store
        return memo[args]  # Return cached value
    
    # Add utility methods to inspect cache
    wrapper.cache = memo
    wrapper.cache_clear = lambda: memo.clear()
    
    return wrapper

@cache_solution
def fib(n):
    if n < 2:
        return n
    return fib(n - 1) + fib(n - 2)

# Demonstrate caching
print("Fibonacci with caching:")
print(f"fib(35) = {fib(35)}")
print(f"Cache contains {len(fib.cache)} entries")
print(f"Cache preview: {dict(list(fib.cache.items())[:5])}...")

# Compare: Without caching, fib(35) would take ~3.5 billion function calls!
# With caching: only 36 calls (one per unique value from 0 to 35)

# Show functools.lru_cache (production-ready version)
from functools import lru_cache

@lru_cache(maxsize=None)  # None = unlimited cache
def fib_lru(n):
    if n < 2:
        return n
    return fib_lru(n - 1) + fib_lru(n - 2)

print(f"\nlru_cache version: fib(100) = {fib_lru(100)}")
print(f"Cache info: {fib_lru.cache_info()}")
````

## 📚 Exercise 6: Data Pipeline Class - Method Chaining & Fluent Interface

### What is Method Chaining?
**Method chaining** (also called **fluent interface**) is a design pattern where methods return `self`, allowing multiple operations to be called in sequence.

```python
# Traditional approach
pipeline = DataPipeline(data)
pipeline.filter(lambda x: x > 0)
pipeline.map(lambda x: x * 2)
result = pipeline.collect()

# Fluent/chained approach
result = DataPipeline(data).filter(lambda x: x > 0).map(lambda x: x * 2).collect()
```

### How Does It Work?
The key is returning `self` from each method:

```python
def filter(self, func):
    self.data = [x for x in self.data if func(x)]
    return self  # This enables chaining!
```

### Why Use Method Chaining?
1. **Readability**: Operations read left-to-right as a "pipeline"
2. **Immutability Option**: Can return new instances for thread safety
3. **Lazy Evaluation**: Can defer computation until `.collect()`

### Real-World Examples
- **Pandas**: `df.dropna().fillna(0).groupby('col').mean()`
- **PySpark**: `df.filter(...).select(...).groupBy(...)`

- **jQuery**: `$('#elem').css('color', 'red').fadeIn()`

````python
# Solution with comprehensive implementation
from functools import reduce as functools_reduce

class DataPipelineSolution:
    def __init__(self, data):
        self.data = list(data)
        self.operations = []  # Track operations for debugging
    
    def filter(self, func):
        """Keep only elements where func(element) is True."""
        self.data = [x for x in self.data if func(x)]
        self.operations.append(f"filter({func.__name__ if hasattr(func, '__name__') else 'lambda'})")
        return self  # Key: return self for chaining
    
    def map(self, func):
        """Transform each element using func."""
        self.data = [func(x) for x in self.data]
        self.operations.append(f"map({func.__name__ if hasattr(func, '__name__') else 'lambda'})")
        return self
    
    def reduce(self, func, initial=None):
        """Combine all elements into single value."""
        if initial is not None:
            return functools_reduce(func, self.data, initial)
        return functools_reduce(func, self.data)
    
    def distinct(self):
        """Remove duplicates (preserves order)."""
        seen = set()
        self.data = [x for x in self.data if not (x in seen or seen.add(x))]
        return self
    
    def sort(self, key=None, reverse=False):
        """Sort elements."""
        self.data.sort(key=key, reverse=reverse)
        return self
    
    def take(self, n):
        """Take first n elements."""
        self.data = self.data[:n]
        return self
    
    def collect(self):
        """Terminal operation: return the data."""
        return self.data
    
    def __repr__(self):
        return f"DataPipeline({self.data})"


# Demo comprehensive usage
print("Example 1: Filter, Map, Collect")
result = (
    DataPipelineSolution(range(10))
    .filter(lambda x: x % 2 == 0)  # Keep even numbers
    .map(lambda x: x ** 2)          # Square them
    .collect()
)
print(f"  Even squares from 0-9: {result}")

print("\nExample 2: Full pipeline with reduce")
total = (
    DataPipelineSolution([1, 2, 3, 4, 5, 5, 4, 3])
    .distinct()                     # Remove duplicates
    .filter(lambda x: x > 2)        # Keep > 2
    .reduce(lambda a, b: a + b)     # Sum them
)
print(f"  Sum of unique values > 2: {total}")

print("\nExample 3: Sort and take")
top3 = (
    DataPipelineSolution([5, 2, 8, 1, 9, 3])
    .sort(reverse=True)             # Sort descending
    .take(3)                        # Top 3
    .collect()
)
print(f"  Top 3 values: {top3}")
````

## 📚 Exercise 7: Valid Parentheses - Stack Data Structure

### What is a Stack?
A **stack** is a Last-In-First-Out (LIFO) data structure. Think of a stack of plates - you can only add or remove from the top.

### Stack Operations
| Operation | Description | Time Complexity |
|-----------|-------------|----------------|
| `push(x)` | Add to top | O(1) |
| `pop()` | Remove from top | O(1) |
| `peek()` | View top (no remove) | O(1) |
| `isEmpty()` | Check if empty | O(1) |

### In Python: Lists as Stacks
```python
stack = []          # Empty stack
stack.append(1)     # Push: [1]
stack.append(2)     # Push: [1, 2]
top = stack.pop()   # Pop: returns 2, stack = [1]
top = stack[-1]     # Peek: returns 1, stack unchanged
```

### Why Use Stack for Matching?
The insight: When we see a closing bracket, it must match the **most recent** opening bracket. This is exactly LIFO behavior!

```
Input: "([{}])"

( → push '('        stack: ['(']
[ → push '['        stack: ['(', '[']
{ → push '{'        stack: ['(', '[', '{']
} → matches '{' ✓   stack: ['(', '[']
] → matches '[' ✓   stack: ['(']
) → matches '(' ✓   stack: []

Stack empty at end → Valid!
```

### Real-World Applications of Stacks
- **Syntax Parsing**: Compilers check code syntax
- **Undo/Redo**: Editor actions stored in stack
- **Browser History**: Back button uses a stack
- **Function Call Stack**: How recursion works internally
- **Expression Evaluation**: Postfix notation calculators

## 📚 Exercise 8: Merge Intervals - Sorting + Greedy Algorithm

### What is the Merge Intervals Problem?
Given a collection of intervals, merge all overlapping intervals.

```
Input:  [[1,3], [2,6], [8,10], [15,18]]
Output: [[1,6], [8,10], [15,18]]
```

### The Greedy Approach
**Greedy algorithms** make locally optimal choices at each step, hoping to find a global optimum. For intervals:

1. **Sort by start time** - ensures we process intervals in order
2. **For each interval**, compare with the last merged interval:
   - If overlaps: extend the end time
   - If no overlap: add as new interval

### When Do Intervals Overlap?
```
Interval A: [a_start, a_end]
Interval B: [b_start, b_end]  (where b_start >= a_start after sorting)

They overlap if: b_start <= a_end

Example 1: [1,3] and [2,6] → 2 <= 3 → OVERLAP, merge to [1,6]
Example 2: [1,3] and [5,7] → 5 > 3 → NO OVERLAP
```

### Why Sort First?
Without sorting, we'd need to compare every interval with every other interval: O(n²)
With sorting, we only need to compare with the previous interval: O(n log n)

- **Network Bandwidth**: Combining data transfer windows
- **Date Range Queries**: Combining overlapping date filters

````python
# Solution with visualization
def merge_intervals_solution(intervals: List[List[int]], debug=False) -> List[List[int]]:
    if not intervals:
        return []
    
    # Sort by start time
    intervals.sort(key=lambda x: x[0])
    if debug: print(f"Sorted: {intervals}")
    
    merged = [intervals[0]]
    if debug: print(f"Initialize with: {merged}")
    
    for start, end in intervals[1:]:
        last_end = merged[-1][1]
        
        if start <= last_end:  # Overlap!
            merged[-1][1] = max(last_end, end)  # Extend
            if debug: print(f"[{start},{end}] overlaps with {merged[-1]} → extend to {merged[-1]}")
        else:  # No overlap
            merged.append([start, end])
            if debug: print(f"[{start},{end}] no overlap → add new interval")
    
    return merged

# Trace example
print("Tracing merge_intervals([[1,3],[2,6],[8,10],[15,18]]):")
result = merge_intervals_solution([[1,3],[2,6],[8,10],[15,18]], debug=True)
print(f"\nResult: {result}")

# Visual representation
def visualize_intervals(intervals, title):
    print(f"\n{title}:")
    max_end = max(end for _, end in intervals) if intervals else 0
    for start, end in intervals:
        line = ' ' * start + '█' * (end - start) + ' ' * (max_end - end + 1)
        print(f"  [{start:2},{end:2}] |{line}|")

original = [[1,3],[2,6],[8,10],[15,18]]
visualize_intervals(original, "Original intervals")
visualize_intervals(merge_intervals_solution(original.copy()), "After merging")
````
