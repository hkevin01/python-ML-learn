## Part 1: Coding Problems (LeetCode Easy/Medium)

```python
# ============================================================
# PROBLEM 1: TWO SUM (Classic!)
# ============================================================
# Given array of integers, return indices of two numbers that add up to target

def two_sum(nums: list, target: int) -> list:
    """
    Find two indices whose values sum to target.
    Time: O(n), Space: O(n)
    """
    seen = {}  # value -> index
    
    for i, num in enumerate(nums):
        complement = target - num
        if complement in seen:
            return [seen[complement], i]
        seen[num] = i
    
    return []

# Test
print("Two Sum:")
print(f"  [2,7,11,15], target=9 → {two_sum([2,7,11,15], 9)}")
print(f"  [3,2,4], target=6 → {two_sum([3,2,4], 6)}")
```

```python
# ============================================================
# PROBLEM 2: VALID PARENTHESES
# ============================================================

def is_valid_parentheses(s: str) -> bool:
    """
    Check if parentheses are balanced.
    Time: O(n), Space: O(n)
    """
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    
    for char in s:
        if char in mapping:
            if not stack or stack.pop() != mapping[char]:
                return False
        else:
            stack.append(char)
    
    return len(stack) == 0

# Test
print("Valid Parentheses:")
print(f"  '()[]{{}}' → {is_valid_parentheses('()[]{}')}")  # True
print(f"  '(]' → {is_valid_parentheses('(]')}")  # False
print(f"  '([)]' → {is_valid_parentheses('([)]')}")  # False
```

```python
# ============================================================
# PROBLEM 3: MERGE SORTED ARRAYS
# ============================================================

def merge_sorted(arr1: list, arr2: list) -> list:
    """
    Merge two sorted arrays into one sorted array.
    Time: O(n+m), Space: O(n+m)
    """
    result = []
    i, j = 0, 0
    
    while i < len(arr1) and j < len(arr2):
        if arr1[i] <= arr2[j]:
            result.append(arr1[i])
            i += 1
        else:
            result.append(arr2[j])
            j += 1
    
    # Add remaining elements
    result.extend(arr1[i:])
    result.extend(arr2[j:])
    
    return result

# Test
print("Merge Sorted Arrays:")
print(f"  [1,3,5] + [2,4,6] → {merge_sorted([1,3,5], [2,4,6])}")
```

```python
# ============================================================
# PROBLEM 4: FIND DUPLICATES
# ============================================================

def find_duplicates(nums: list) -> list:
    """
    Find all duplicates in array.
    Time: O(n), Space: O(n)
    """
    from collections import Counter
    counts = Counter(nums)
    return [num for num, count in counts.items() if count > 1]

# Test
print("Find Duplicates:")
print(f"  [1,2,3,2,1,4] → {find_duplicates([1,2,3,2,1,4])}")
```

```python
# ============================================================
# PROBLEM 5: LONGEST SUBSTRING WITHOUT REPEATING
# ============================================================

def longest_substring(s: str) -> int:
    """
    Find length of longest substring without repeating characters.
    Sliding window approach.
    Time: O(n), Space: O(min(m,n)) where m is charset size
    """
    char_index = {}  # char -> last seen index
    max_length = 0
    start = 0
    
    for i, char in enumerate(s):
        if char in char_index and char_index[char] >= start:
            start = char_index[char] + 1
        else:
            max_length = max(max_length, i - start + 1)
        
        char_index[char] = i
    
    return max_length

# Test
print("Longest Substring Without Repeating:")
print(f"  'abcabcbb' → {longest_substring('abcabcbb')}")  # 3 (abc)
print(f"  'bbbbb' → {longest_substring('bbbbb')}")  # 1
```

```python
# ============================================================
# PROBLEM 6: REVERSE LINKED LIST
# ============================================================

class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def reverse_linked_list(head: ListNode) -> ListNode:
    """
    Reverse a linked list in place.
    Time: O(n), Space: O(1)
    """
    prev = None
    current = head
    
    while current:
        next_temp = current.next
        current.next = prev
        prev = current
        current = next_temp
    
    return prev

# Helper to create and print list
def list_to_linked(arr):
    dummy = ListNode()
    current = dummy
    for val in arr:
        current.next = ListNode(val)
        current = current.next
    return dummy.next

def linked_to_list(head):
    result = []
    while head:
        result.append(head.val)
        head = head.next
    return result

# Test
head = list_to_linked([1, 2, 3, 4, 5])
reversed_head = reverse_linked_list(head)
print(f"Reverse Linked List: [1,2,3,4,5] → {linked_to_list(reversed_head)}")
```

## Part 2: Pandas Operations (Data Manipulation)

```python
import pandas as pd
import numpy as np

# Create sample datasets
np.random.seed(42)

# Orders data
orders = pd.DataFrame({
    'order_id': range(1, 11),
    'customer_id': [1, 2, 1, 3, 2, 1, 4, 3, 2, 1],
    'product': ['A', 'B', 'A', 'C', 'A', 'B', 'A', 'B', 'C', 'A'],
    'amount': np.random.randint(10, 100, 10),
    'date': pd.date_range('2024-01-01', periods=10, freq='D')
})

# Customers data
customers = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 5],
    'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'city': ['NYC', 'LA', 'NYC', 'Chicago', 'LA']
})

print("Orders:")
print(orders.head())
print("\nCustomers:")
print(customers)
```

```python
# ============================================================
# TASK 1: GROUPBY & AGGREGATION
# ============================================================
# Find total and average order amount by customer

customer_stats = orders.groupby('customer_id').agg({
    'amount': ['sum', 'mean', 'count'],
    'order_id': 'nunique'
}).round(2)

customer_stats.columns = ['total_amount', 'avg_amount', 'num_orders', 'unique_orders']
print("Customer Statistics:")
print(customer_stats)
```

```python
# ============================================================
# TASK 2: JOINS
# ============================================================
# Join orders with customers

# Inner join (only matching rows)
inner = orders.merge(customers, on='customer_id', how='inner')
print("Inner Join (first 5 rows):")
print(inner.head())

# Left join (all orders, even if no customer match)
left = orders.merge(customers, on='customer_id', how='left')

# Find customers who never ordered
all_customers = customers.merge(orders, on='customer_id', how='left')
never_ordered = all_customers[all_customers['order_id'].isna()]['name'].unique()
print(f"\nCustomers who never ordered: {list(never_ordered)}")
```

```python
# ============================================================
# TASK 3: WINDOW FUNCTIONS
# ============================================================
# Calculate running total and rank within customer

orders_ranked = orders.copy()

# Running total by customer
orders_ranked['running_total'] = orders_ranked.groupby('customer_id')['amount'].cumsum()

# Rank orders by amount within each customer
orders_ranked['rank_by_amount'] = orders_ranked.groupby('customer_id')['amount'].rank(ascending=False)

print("Orders with Window Functions:")
print(orders_ranked[['order_id', 'customer_id', 'amount', 'running_total', 'rank_by_amount']])
```

```python
# ============================================================
# TASK 4: PIVOT & RESHAPE
# ============================================================
# Pivot: product sales by customer

pivot = orders.pivot_table(
    values='amount',
    index='customer_id',
    columns='product',
    aggfunc='sum',
    fill_value=0
)

print("Pivot Table (Customer x Product):")
print(pivot)
```

```python
# ============================================================
# TASK 5: FILTERING & CONDITIONS
# ============================================================

# Multiple conditions
high_value_nyc = inner[
    (inner['amount'] > 50) & 
    (inner['city'] == 'NYC')
]
print("High value orders from NYC:")
print(high_value_nyc[['order_id', 'name', 'amount', 'city']])

# Using query (SQL-like syntax)
same_result = inner.query("amount > 50 and city == 'NYC'")
print("\nSame result using query():")
print(same_result[['order_id', 'name', 'amount', 'city']])
```

## Part 3: SQL Equivalents

```python
# ============================================================
# PANDAS TO SQL TRANSLATION GUIDE
# ============================================================

print("📋 Pandas ↔ SQL Translation Guide:")
print("="*70)

translations = [
    {
        'operation': 'Select columns',
        'sql': "SELECT col1, col2 FROM table",
        'pandas': "df[['col1', 'col2']]"
    },
    {
        'operation': 'Filter rows',
        'sql': "SELECT * FROM table WHERE col > 10",
        'pandas': "df[df['col'] > 10]"
    },
    {
        'operation': 'Group by + aggregate',
        'sql': "SELECT col, SUM(val) FROM table GROUP BY col",
        'pandas': "df.groupby('col')['val'].sum()"
    },
    {
        'operation': 'Having clause',
        'sql': "SELECT col, COUNT(*) FROM t GROUP BY col HAVING COUNT(*) > 5",
        'pandas': "df.groupby('col').filter(lambda x: len(x) > 5)"
    },
    {
        'operation': 'Inner join',
        'sql': "SELECT * FROM a JOIN b ON a.id = b.id",
        'pandas': "a.merge(b, on='id', how='inner')"
    },
    {
        'operation': 'Left join',
        'sql': "SELECT * FROM a LEFT JOIN b ON a.id = b.id",
        'pandas': "a.merge(b, on='id', how='left')"
    },
    {
        'operation': 'Order by',
        'sql': "SELECT * FROM table ORDER BY col DESC",
        'pandas': "df.sort_values('col', ascending=False)"
    },
    {
        'operation': 'Limit',
        'sql': "SELECT * FROM table LIMIT 10",
        'pandas': "df.head(10)"
    },
    {
        'operation': 'Distinct',
        'sql': "SELECT DISTINCT col FROM table",
        'pandas': "df['col'].unique()"
    },
    {
        'operation': 'Count distinct',
        'sql': "SELECT COUNT(DISTINCT col) FROM table",
        'pandas': "df['col'].nunique()"
    },
    {
        'operation': 'Case when',
        'sql': "SELECT CASE WHEN col > 10 THEN 'high' ELSE 'low' END",
        'pandas': "np.where(df['col'] > 10, 'high', 'low')"
    },
    {
        'operation': 'Window function (rank)',
        'sql': "SELECT *, RANK() OVER (PARTITION BY grp ORDER BY val)",
        'pandas': "df.groupby('grp')['val'].rank()"
    }
]

for t in translations:
    print(f"\n{t['operation']}:")
    print(f"  SQL:    {t['sql']}")
    print(f"  Pandas: {t['pandas']}")
```

## Part 4: Explaining Projects (STAR Method)

```python
# ============================================================
# THE STAR METHOD FOR PROJECT EXPLANATION
# ============================================================

print("📋 STAR Method for Explaining Projects:")
print("="*60)
print("""
S - SITUATION
=============
What was the business problem or context?
  - "Our team needed to reduce customer churn, which was costing $X/month"
  - "The company wanted to automate manual review of 10K documents/day"

T - TASK
========
What was YOUR specific responsibility?
  - "I was responsible for building the ML model"
  - "I led the data preprocessing and feature engineering"
  - "I owned the entire pipeline from data collection to deployment"

A - ACTION
==========
What did you DO? Be specific and technical!
  - "I trained a Random Forest classifier on 50K samples"
  - "I used TF-IDF with logistic regression for text classification"
  - "I implemented a real-time scoring API using FastAPI"

R - RESULT
==========
What was the IMPACT? Quantify!
  - "Improved F1 score from 0.65 to 0.82"
  - "Reduced churn by 15%, saving $500K annually"
  - "Decreased processing time from 2 hours to 5 minutes"
""")
```

```python
# ============================================================
# COMMUNICATION TIPS
# ============================================================

print("📋 Communication Tips for Interviews:")
print("="*60)
print("""
✅ DO:
  - Start with business context, then get technical
  - Quantify impact (numbers, percentages, dollars)
  - Acknowledge trade-offs and alternatives you considered
  - Be honest about what didn't work
  - Ask clarifying questions
  - Think out loud when problem-solving

❌ DON'T:
  - Jump into implementation without understanding the problem
  - Use jargon without explaining it
  - Claim sole credit for team projects
  - Say "I don't know" without trying
  - Oversell or lie about impact

📌 WHEN STUCK:
  - "Let me think about this for a moment..."
  - "I haven't done exactly this, but my approach would be..."
  - "Can you give me a hint about which direction you're looking for?"
  - "Let me start with a brute force approach and optimize from there"
""")
```
