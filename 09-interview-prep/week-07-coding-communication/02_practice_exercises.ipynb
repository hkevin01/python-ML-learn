```python
# Solution
def top_k_frequent_solution(nums: List[int], k: int) -> List[int]:
    # Count frequencies
    count = Counter(nums)
    
    # Use min heap to keep track of top k
    # Heap stores (frequency, num)
    heap = []
    
    for num, freq in count.items():
        heapq.heappush(heap, (freq, num))
        if len(heap) > k:
            heapq.heappop(heap)
    
    return [num for freq, num in heap]


# Alternative: Bucket Sort O(n)
def top_k_frequent_bucket(nums: List[int], k: int) -> List[int]:
    count = Counter(nums)
    
    # Bucket: index = frequency, value = list of nums with that freq
    buckets = [[] for _ in range(len(nums) + 1)]
    for num, freq in count.items():
        buckets[freq].append(num)
    
    # Collect from highest frequency buckets
    result = []
    for i in range(len(buckets) - 1, 0, -1):
        result.extend(buckets[i])
        if len(result) >= k:
            return result[:k]
    
    return result

# Test both solutions
print("Heap solution:", top_k_frequent_solution([1,1,1,2,2,3], 2))
print("Bucket solution:", top_k_frequent_bucket([1,1,1,2,2,3], 2))
```

```python
# Solution
def product_except_self_solution(nums: List[int]) -> List[int]:
    n = len(nums)
    result = [1] * n
    
    # First pass: calculate prefix products
    # result[i] = product of all elements to the left of i
    prefix = 1
    for i in range(n):
        result[i] = prefix
        prefix *= nums[i]
    
    # Second pass: multiply by suffix products
    # result[i] *= product of all elements to the right of i
    suffix = 1
    for i in range(n - 1, -1, -1):
        result[i] *= suffix
        suffix *= nums[i]
    
    return result

# Walk through example
nums = [1, 2, 3, 4]
print(f"Input: {nums}")
print(f"Output: {product_except_self_solution(nums)}")
print("\nExplanation:")
print("  [24, 12, 8, 6]")
print("  24 = 2*3*4, 12 = 1*3*4, 8 = 1*2*4, 6 = 1*2*3")
```

```python
# Solution using OrderedDict
from collections import OrderedDict

class LRUCacheSolution:
    def __init__(self, capacity: int):
        self.capacity = capacity
        self.cache = OrderedDict()
    
    def get(self, key: int) -> int:
        if key not in self.cache:
            return -1
        # Move to end (most recently used)
        self.cache.move_to_end(key)
        return self.cache[key]
    
    def put(self, key: int, value: int) -> None:
        if key in self.cache:
            # Update and move to end
            self.cache.move_to_end(key)
            self.cache[key] = value
        else:
            # Insert new key
            if len(self.cache) >= self.capacity:
                # Evict LRU (first item)
                self.cache.popitem(last=False)
            self.cache[key] = value


# More detailed test
cache = LRUCacheSolution(2)
cache.put(1, 1)
print(f"After put(1,1): {dict(cache.cache)}")
cache.put(2, 2)
print(f"After put(2,2): {dict(cache.cache)}")
print(f"get(1) = {cache.get(1)}")
print(f"After get(1): {dict(cache.cache)}")
cache.put(3, 3)  # Evicts 2
print(f"After put(3,3): {dict(cache.cache)}")
print(f"get(2) = {cache.get(2)} (evicted)")
```

```python
# Solution
def length_of_longest_substring_solution(s: str) -> int:
    char_index = {}  # char -> last seen index
    max_length = 0
    left = 0
    
    for right, char in enumerate(s):
        # If char was seen and is within current window
        if char in char_index and char_index[char] >= left:
            # Move left pointer past the duplicate
            left = char_index[char] + 1
        
        # Update last seen index
        char_index[char] = right
        
        # Update max length
        max_length = max(max_length, right - left + 1)
    
    return max_length

# Walk through example
s = "abcabcbb"
print(f"Input: '{s}'")
print(f"Output: {length_of_longest_substring_solution(s)}")
print("\nStep by step:")
print("  a -> window: 'a', length: 1")
print("  b -> window: 'ab', length: 2")
print("  c -> window: 'abc', length: 3 (max)")
print("  a -> duplicate! window: 'bca', length: 3")
print("  b -> duplicate! window: 'cab', length: 3")
print("  ...")
```
