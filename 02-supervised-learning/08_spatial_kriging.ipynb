# Spatial Regression: Kriging — A Gentle Intro

## What it is
Kriging is a geostatistical interpolation method that predicts values at unknown locations using spatial autocorrelation (nearby points are similar).

## What it does
- Uses a variogram model to weight nearby samples for interpolation.
- Produces both an estimate and an uncertainty (kriging variance).

## When to use
- You have scattered spatial measurements and want a smooth surface.
- You care about uncertainty maps as much as the estimates.

```python
# Imports
import numpy as np  # numerical arrays
import matplotlib.pyplot as plt  # plotting
from pykrige.ok import OrdinaryKriging  # Ordinary Kriging interpolator

np.random.seed(42)  # reproducibility
n = 80  # number of samples
x = np.random.uniform(0, 100, n)  # sample X-coordinates
y = np.random.uniform(0, 100, n)  # sample Y-coordinates
z_true = 0.12*x + 0.06*y + 8*np.sin(x/18) + 6*np.cos(y/24)  # underlying smooth field
z = z_true + np.random.normal(0, 1.8, n)  # add measurement noise
print(f'Samples: {n}')  # quick info
```

```python
# Ordinary Kriging interpolation over a grid
gridx = np.linspace(0, 100, 80)  # grid X-coordinates
gridy = np.linspace(0, 100, 80)  # grid Y-coordinates
OK = OrdinaryKriging(x, y, z, variogram_model='spherical', verbose=False, enable_plotting=False)  # set model
zgrid, ss = OK.execute('grid', gridx, gridy)  # interpolate and get variance

fig, axs = plt.subplots(1, 2, figsize=(12, 5))  # two plots: estimate + variance
im0 = axs[0].imshow(zgrid.T, origin='lower', extent=(gridx.min(), gridx.max(), gridy.min(), gridy.max()), cmap='viridis')  # estimated surface
axs[0].scatter(x, y, c=z, s=20, edgecolor='k', linewidth=0.5)  # sample points
axs[0].set_title('Ordinary Kriging: Interpolated Field')  # title
fig.colorbar(im0, ax=axs[0], fraction=0.046, pad=0.04)  # colorbar

im1 = axs[1].imshow(ss.T, origin='lower', extent=(gridx.min(), gridx.max(), gridy.min(), gridy.max()), cmap='magma')  # kriging variance
axs[1].set_title('Kriging Variance (Uncertainty)')  # title
fig.colorbar(im1, ax=axs[1], fraction=0.046, pad=0.04)  # colorbar
plt.tight_layout()  # tidy layout
plt.show()  # render
```

### Key ideas
- Ordinary Kriging assumes a constant mean; weights depend on the variogram and distances.
- The kriging variance maps interpolation uncertainty (higher far from samples).
- Choice of variogram (spherical, exponential, etc.) strongly influences results.
